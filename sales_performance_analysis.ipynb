```python
import pandas as pd

data = {
    "Employee": ["Ali", "Sara", "Ahmed", "Ayesha", "Usman", "Zara", "Bilal", "Hira"],
    "Department": ["Sales", "Sales", "HR", "Sales", "HR", "Sales", "HR", "Sales"],
    "Sales": [5000, 7000, None, 8000, 4500, 6000, None, 7500],
    "Region": ["North", "South", "North", "East", "South", "East", "North", "South"]
}

df = pd.DataFrame(data)

# Save as CSV file
df.to_csv("sales_data.csv", index=False)

df
```

```python
df = pd.read_csv("sales_data.csv")
df.head()
```

```python
df["Sales"] = df["Sales"].fillna(df["Sales"].mean())
df.isnull().sum()
```

```python
df.groupby("Department")["Sales"].mean()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.barplot(x="Employee", y="Sales", data=df)
plt.xticks(rotation=45)
plt.show()
```

## Insights

- Missing values were handled using mean
- Sales department performs better
- Ayesha is top performer
